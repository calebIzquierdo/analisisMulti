```python
# 3. Descomprimir el dataset
import zipfile
import os

zip_path = '/content/drive/MyDrive/2026/upeu-2026-1/Investigacion 4/dataset-entrenamiento/dataset-v2.zip'

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall('/content/dataset-v2')
```

```python
# 4. Cargar modelo entrenado
from ultralytics import YOLO

model = YOLO(
    '/content/drive/MyDrive/yolov8_runs/arroz_v2/weights/best.pt'
)
```

```python
# 5. Evaluar con el conjunto TEST
metrics = model.val(
    data='/content/dataset-v2/data.yaml',
    split='test',
    imgsz=640,
    device='cpu',
    plots=True,
    project='/content/evaluacion_final',
    name='test_arroz',
    exist_ok=True
)
```

```python
# 6. Mostrar matriz de confusión
from IPython.display import Image, display

display(Image(
    '/content/evaluacion_final/test_arroz/confusion_matrix_normalized.png',
    width=900
))
```

```python
# 7. Mostrar métricas
print("mAP@0.5:", metrics.box.map50)
print("mAP@0.5:0.95:", metrics.box.map)
print("Precisión:", metrics.box.mp)
print("Recall:", metrics.box.mr)
```
